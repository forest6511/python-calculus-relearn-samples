## LESSON 06 「平均の速さ」は、2点を結ぶ直線の傾き

本書の「LESSON 06 「平均の速さ」は、2点を結ぶ直線の傾き」の説明を読みながら、下のコードを ▶ で試してください。

```python
# 発進から t 秒後までに進んだ距離(m)
def kyori(t):
    return t ** 2

# 平均の速さ = 進んだ距離 ÷ かかった時間 (2点を結ぶ直線の傾き)
def heikin_hayasa(t1, t2):
    return (kyori(t2) - kyori(t1)) / (t2 - t1)

print(f"0秒から10秒の平均: 秒速{heikin_hayasa(0, 10)}m")
print(f"0秒から5秒の平均: 秒速{heikin_hayasa(0, 5)}m")
print(f"5秒から10秒の平均: 秒速{heikin_hayasa(5, 10)}m")

print(f"9秒から10秒の平均: 秒速{heikin_hayasa(9, 10)}m")
```

## LESSON 07 区間を狭めて、「瞬間の速さ」を追いつめる

本書の「LESSON 07 区間を狭めて、「瞬間の速さ」を追いつめる」の説明を読みながら、下のコードを ▶ で試してください。

```python
# 発進から t 秒後までに進んだ距離(m)
def kyori(t):
    return t ** 2

# 平均の速さ (2点を結ぶ直線の傾き)
def heikin_hayasa(t1, t2):
    return (kyori(t2) - kyori(t1)) / (t2 - t1)

# 5秒後から、幅 h だけあとまでの平均を、h を縮めながら見る
for h in [1, 0.5, 0.1, 0.01, 0.001]:
    v = heikin_hayasa(5, 5 + h)
    print(f"5秒から{5 + h}秒まで: 秒速{v:.3f}m")

for h in [0.1, 0.01, 0.001]:
    v = heikin_hayasa(5 - h, 5)
    print(f"{5 - h}秒から5秒まで: 秒速{v:.3f}m")
```

## LESSON 08 接線は、グラフの「その場の向き」

本書の「LESSON 08 接線は、グラフの「その場の向き」」の説明を読みながら、下のコードを ▶ で試してください。

```python
def f(x):
    return x ** 2

# x = a での瞬間の傾き(h を小さくとった平均変化率で近似)
def shunkan_katamuki(a, h=0.001):
    return (f(a + h) - f(a)) / h

a = 1                          # 接する場所
katamuki_a = shunkan_katamuki(a)

x = np.linspace(-3, 3, 100)
# 点(a, f(a)) を通る、傾き katamuki_a の直線
sessen = katamuki_a * (x - a) + f(a)

plt.plot(x, f(x), label="y = x の2乗")
plt.plot(x, sessen, label="x = 1 での接線")
plt.scatter([a], [f(a)], color="red", zorder=3)  # 接する点に印
plt.xlabel("x")
plt.ylabel("y")
plt.title("放物線と接線")
plt.legend()
plt.grid(True)
plt.show()

print(f"x = {a} での瞬間の傾き: {katamuki_a:.3f}")

a = -1   # 接する場所を左半分に移した
```

## LESSON 09 てっぺんで傾きゼロ、そして「微分する」の意味

本書の「LESSON 09 てっぺんで傾きゼロ、そして「微分する」の意味」の説明を読みながら、下のコードを ▶ で試してください。

```python
# 投げ上げてから t 秒後のボールの高さ(m)
def takasa(t):
    return 20 * t - 5 * t ** 2

# t 秒後での瞬間の傾き(h を小さくとった平均変化率で近似)
def katamuki_t(t, h=0.001):
    return (takasa(t + h) - takasa(t)) / h

t = np.linspace(0, 4, 100)
plt.plot(t, takasa(t))
plt.xlabel("時間(秒)")
plt.ylabel("高さ(m)")
plt.title("投げ上げたボールの高さ")
plt.grid(True)
plt.show()

for ti in [1, 2, 3]:
    print(f"t = {ti}秒での傾き: {katamuki_t(ti):.3f}")

def takasa(t):
    return 30 * t - 5 * t ** 2   # 初速を秒速30mに上げた
```
