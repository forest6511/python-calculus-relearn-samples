## LESSON 29 速度の記録だけで、進んだ距離は分かるか

本書の「LESSON 29 速度の記録だけで、進んだ距離は分かるか」の説明を読みながら、下のコードを ▶ で試してください。

```python
def sokudo(t):
    return 3 * t                    # t 秒後の速度(m/秒)。10秒後に30

haba = 1.0                          # 短冊の幅(秒)
t = np.arange(0, 10, haba)          # 各短冊の左端の時刻
kyori = np.sum(sokudo(t) * haba)    # 高さ × 幅 を全部足す
print(f"短冊の合計: {kyori:.1f} m")

t_hidari = np.arange(0, 10, 1.0)    # 各短冊の左端(0, 1, ..., 9秒)

plt.bar(t_hidari, sokudo(t_hidari), width=1.0, align="edge",
        color="lightgray", edgecolor="black", label="短冊(高さ×幅)")
t = np.linspace(0, 10, 101)
plt.plot(t, sokudo(t), label="速度 3t")
plt.xlabel("時間(秒)")
plt.ylabel("速度(m/秒)")
plt.title("速度のグラフの下を短冊で埋める")
plt.legend()
plt.grid(True)
plt.show()

t_migi = np.arange(1, 11, haba)         # 各短冊の右端の時刻
kyori_migi = np.sum(sokudo(t_migi) * haba)
print(f"右端ではかった合計: {kyori_migi:.1f} m")
```

## LESSON 30 短冊を細くしていくと、何が起きるか

本書の「LESSON 30 短冊を細くしていくと、何が起きるか」の説明を読みながら、下のコードを ▶ で試してください。

```python
for haba in [1.0, 0.1, 0.01, 0.001]:
    t_hidari = np.arange(0, 10, haba)
    hidari = np.sum(sokudo(t_hidari) * haba)
    migi = np.sum(sokudo(t_hidari + haba) * haba)
    print(f"幅 {haba:>6}: 左端 {hidari:>8.3f}   右端 {migi:>8.3f}")

def nijou(x):
    return x ** 2

for haba in [0.001, 0.0001]:
    x_hidari = np.arange(0, 1, haba)
    menseki = np.sum(nijou(x_hidari) * haba)
    print(f"幅 {haba:>7}: 面積 {menseki:.5f}")
```

## LESSON 31 短冊1万本の代わりに、形を変えてみる

本書の「LESSON 31 短冊1万本の代わりに、形を変えてみる」の説明を読みながら、下のコードを ▶ で試してください。

```python
haba = 0.01
x_hidari = np.arange(0, 1, haba)
tanzaku = np.sum(nijou(x_hidari) * haba)

x = np.linspace(0, 1, 101)
daikei = np.trapezoid(nijou(x), x)

print(f"短冊100本: {tanzaku:.5f}   真の値 1/3 とのずれ {abs(tanzaku - 1/3):.5f}")
print(f"台形100個: {daikei:.5f}   真の値 1/3 とのずれ {abs(daikei - 1/3):.5f}")

t = np.linspace(0, 10, 11)          # 1秒おき、たった11点
kyori = np.trapezoid(sokudo(t), t)
print(f"台形則で出した距離: {kyori:.1f} m")
```

## LESSON 32 行って戻ると、面積はどうなるか

本書の「LESSON 32 行って戻ると、面積はどうなるか」の説明を読みながら、下のコードを ▶ で試してください。

```python
def bouru(t):
    return 20 - 10 * t     # 第2章のボールの速度(m/秒)

t_nobori = np.linspace(0, 2, 201)   # 上がっている間
t_kudari = np.linspace(2, 4, 201)   # 落ちている間
t_zenbu  = np.linspace(0, 4, 401)

print(f"前半(0〜2秒)の面積: {np.trapezoid(bouru(t_nobori), t_nobori):+.1f}")
print(f"後半(2〜4秒)の面積: {np.trapezoid(bouru(t_kudari), t_kudari):+.1f}")
print(f"全体(0〜4秒)の面積: {np.trapezoid(bouru(t_zenbu), t_zenbu):+.1f}")

t = np.linspace(0, 4, 401)
takasa = []                          # 積み上げた面積を入れていくリスト
for k in range(len(t)):
    made = t[: k + 1]                # 先頭から k+1 個目までの時刻
    takasa.append(np.trapezoid(bouru(made), made))

plt.plot(t, takasa, label="速度の面積の積み上げ")
plt.plot(t, 20 * t - 5 * t ** 2, linestyle="--", label="第2章の高さの式")
plt.xlabel("時間(秒)")
plt.ylabel("高さ(m)")
plt.title("面積を積み上げると高さのグラフに戻る")
plt.legend()
plt.grid(True)
plt.show()

michinori = np.trapezoid(np.abs(bouru(t_zenbu)), t_zenbu)
print(f"道のり: {michinori:.1f} m")
```

## LESSON 33 曲がった図形の面積は、どうやって測るか

本書の「LESSON 33 曲がった図形の面積は、どうやって測るか」の説明を読みながら、下のコードを ▶ で試してください。

```python
def hanken(x):
    return np.sqrt(1 - x ** 2)      # 半径1の円の上半分

x = np.linspace(-1, 1, 401)

plt.fill_between(x, hanken(x), color="lightgray")
plt.plot(x, hanken(x))
plt.axis("equal")                   # 縦横の目盛りをそろえる
plt.xlabel("x")
plt.ylabel("y")
plt.title("半径1の円の上半分")
plt.grid(True)
plt.show()

hanbun = np.trapezoid(hanken(x), x)
print(f"上半分の面積: {hanbun:.4f}")
print(f"円全体の面積: {2 * hanbun:.4f}")
print(f"円周率 π    : {np.pi:.4f}")

basho = np.linspace(0, 14, 8)       # 幅を測った場所(2mおき)
ike_haba = np.array([0, 3.2, 4.5, 5.1, 4.8, 3.9, 2.2, 0])   # 池の幅(m)
menseki = np.trapezoid(ike_haba, basho)
print(f"池のおよその面積: {menseki:.1f} 平方m")

def hanken2(x):
    return np.sqrt(4 - x ** 2)      # 半径2の円の上半分

x2 = np.linspace(-2, 2, 801)
zentai = 2 * np.trapezoid(hanken2(x2), x2)
print(f"半径2の円の面積: {zentai:.2f}")
print(f"π × 2 × 2      : {np.pi * 4:.2f}")
```
