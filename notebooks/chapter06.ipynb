## LESSON 24 山のてっぺんと谷底は、どうすれば探せるか

本書の「LESSON 24 山のてっぺんと谷底は、どうすれば探せるか」の説明を読みながら、下のコードを ▶ で試してください。

```python
# 傾きの測定器(第5章と同じ・この章でも使い回す)
def doukansuu(f, x, h=0.001):
    return (f(x + h) - f(x - h)) / (2 * h)

def yamatani(x):
    return x ** 3 - 3 * x

for a in [-2.0, -1.5, -1.0, -0.5, 0.0, 0.5, 1.0, 1.5, 2.0]:
    print(f"x = {a:+.1f} での傾き: {doukansuu(yamatani, a):+8.4f}")

x = np.linspace(-2.2, 2.2, 89)

plt.plot(x, yamatani(x), label="もとの関数")
plt.plot(x, doukansuu(yamatani, x), linestyle="--", label="各点の傾き")
plt.axhline(0, color="gray", linewidth=1)
plt.xlabel("x")
plt.ylabel("y")
plt.title("山と谷で、傾きのグラフは高さ 0 の線を横切る")
plt.legend()
plt.grid(True)
plt.show()

def sanjou(x):
    return x ** 3

for a in [-1.0, 0.0, 1.0]:
    print(f"x = {a:+.1f} での傾き: {doukansuu(sanjou, a):+8.4f}")
```

## LESSON 25 いちばん大きい箱は、どの切り方か

本書の「LESSON 25 いちばん大きい箱は、どの切り方か」の説明を読みながら、下のコードを ▶ で試してください。

```python
def taiseki(x):
    return x * (20 - 2 * x) ** 2    # 高さ x × 底面 (20 - 2x) 四方

for kiritori in [1, 2, 3, 4, 5]:
    print(f"{kiritori} cm 切り取る: 体積 {taiseki(kiritori):>4} 立方cm")

x = np.linspace(0, 10, 1001)

plt.plot(x, taiseki(x))
plt.xlabel("切り取る正方形の一辺(cm)")
plt.ylabel("箱の体積(立方cm)")
plt.title("切り取りの大きさと箱の体積")
plt.grid(True)
plt.show()

v = taiseki(x)                    # 1001個の候補ぜんぶの体積
ichiban = np.argmax(v)            # いちばん大きい体積は先頭から何番目か
print(f"いちばん大きい体積: {v[ichiban]:.1f} 立方cm")
print(f"そのときの切り取り: {x[ichiban]:.2f} cm")

def taiseki_30(x):
    return x * (30 - 2 * x) ** 2

x30 = np.linspace(0, 15, 1501)
v30 = taiseki_30(x30)
ichiban = np.argmax(v30)
print(f"いちばん大きい体積: {v30[ichiban]:.1f} 立方cm")
print(f"そのときの切り取り: {x30[ichiban]:.2f} cm")
```

## LESSON 26 もう1個作ると、利益はいくら増えるか

本書の「LESSON 26 もう1個作ると、利益はいくら増えるか」の説明を読みながら、下のコードを ▶ で試してください。

```python
def rieki(x):
    return -0.1 * x ** 2 + 30 * x - 500    # 1日 x 個作るときの利益(円)

for kosuu in [50, 100, 150, 200]:
    tsugi = rieki(kosuu + 1) - rieki(kosuu)    # もう1個作ると増える利益
    bibun = doukansuu(rieki, kosuu)
    print(f"{kosuu:>3}個の次の1個: {tsugi:+6.2f}円   微分: {bibun:+6.2f}円")

x = np.linspace(0, 300, 301)

plt.plot(x, rieki(x))
plt.axvline(150, linestyle="--", color="gray")
plt.axhline(0, color="gray", linewidth=1)
plt.xlabel("1日に作る個数")
plt.ylabel("利益(円)")
plt.title("作る個数と利益")
plt.grid(True)
plt.show()

def rieki_yachin(x):
    return -0.1 * x ** 2 + 30 * x - 800    # 家賃が300円ぶん上がった

for kosuu in [149, 150, 151]:
    print(f"{kosuu}個: 利益 {rieki_yachin(kosuu):.2f}円   "
          f"微分 {doukansuu(rieki_yachin, kosuu):+.2f}円")
```

## LESSON 27 投げ上げたボールの、速さの変わり方

本書の「LESSON 27 投げ上げたボールの、速さの変わり方」の説明を読みながら、下のコードを ▶ で試してください。

```python
def takasa(t):
    return 20 * t - 5 * t ** 2     # 第2章のボールの高さ(m)

def sokudo(t):
    return doukansuu(takasa, t)    # 高さの傾き = 速度

for t in [0, 1, 2, 3, 4]:
    kasoku = doukansuu(sokudo, t)          # 速度の傾き
    print(f"t = {t}秒: 速度 {sokudo(t):+7.2f}   速度の傾き {kasoku:+7.2f}")

def takasa_tsuyoku(t):
    return 30 * t - 5 * t ** 2     # 秒速30mで強く投げ上げる

def sokudo_tsuyoku(t):
    return doukansuu(takasa_tsuyoku, t)

for t in [0, 1, 2, 3]:
    kasoku = doukansuu(sokudo_tsuyoku, t)
    print(f"t = {t}秒: 速度 {sokudo_tsuyoku(t):+7.2f}   速度の傾き {kasoku:+7.2f}")
```

## LESSON 28 伸びの勢いが変わる日を、見つけられるか

本書の「LESSON 28 伸びの勢いが変わる日を、見つけられるか」の説明を読みながら、下のコードを ▶ で試してください。

```python
hi = np.linspace(0, 90, 91)                       # 公開からの日数
ruikei = 90000 / (1 + np.exp(-0.1 * (hi - 45)))   # 累計再生回数(練習用データ)

plt.plot(hi, ruikei)
plt.xlabel("公開からの日数")
plt.ylabel("累計再生回数")
plt.title("動画の累計再生回数")
plt.grid(True)
plt.show()

nobi = np.gradient(ruikei, hi)     # 1日あたりの伸び(第4章の道具)
ichiban = np.argmax(nobi)
print(f"伸びがいちばん大きい日: {hi[ichiban]:.0f}日目")
print(f"その日の伸び: 1日あたり {nobi[ichiban]:.0f}回")

plt.plot(hi, nobi)
plt.axvline(45, linestyle="--", color="gray")
plt.xlabel("公開からの日数")
plt.ylabel("1日あたりの伸び(回)")
plt.title("伸びのグラフは山型になる")
plt.grid(True)
plt.show()

def katamuki(x):
    return doukansuu(yamatani, x)   # LESSON 24 の曲線の、各点の傾き

for a in [-1.0, -0.5, 0.0, 0.5, 1.0]:
    print(f"x = {a:+.1f}: 傾きの傾き {doukansuu(katamuki, a):+8.4f}")
```
