## LESSON 14 傾きの測定器を、自分の手で作る

本書の「LESSON 14 傾きの測定器を、自分の手で作る」の説明を読みながら、下のコードを ▶ で試してください。

```python
# どんな関数でも使える、傾きの測定器
def bibun(f, a, h=0.001):
    return (f(a + h) - f(a)) / h    # 2点の差を、幅で割る

# 第2章の車: 進んだ距離
def kyori(t):
    return t ** 2

# 第2章のボール: 高さ
def takasa(t):
    return 20 * t - 5 * t ** 2

# 新顔: 3乗の関数
def sanjou(x):
    return x ** 3

print(f"車の 5秒後の速さ: 秒速{bibun(kyori, 5):.3f}m")
print(f"ボールの 1秒後の傾き: {bibun(takasa, 1):.3f}")
print(f"3乗の関数の x = 2 での傾き: {bibun(sanjou, 2):.3f}")

# 自分で作った関数も、そのまま渡せる
def gyaku(x):
    return 1 / x       # 逆数の関数

print(f"逆数の関数の x = 2 での傾き: {bibun(gyaku, 2):.3f}")
```

## LESSON 15 傾きは、両側から挟んで測ると正確になる

本書の「LESSON 15 傾きは、両側から挟んで測ると正確になる」の説明を読みながら、下のコードを ▶ で試してください。

```python
def sanjou(x):
    return x ** 3

# 前進差分: いまの点と、少し先の点で測る(第2章からのやり方)
def zenshin(f, a, h):
    return (f(a + h) - f(a)) / h

# 中心差分: 少し手前の点と、少し先の点で測る
def chuushin(f, a, h):
    return (f(a + h) - f(a - h)) / (2 * h)

# x = 2 での傾き(本当の値は 12)で、2つの測り方を競わせる
for h in [0.1, 0.01, 0.001]:
    z = zenshin(sanjou, 2, h)
    c = chuushin(sanjou, 2, h)
    print(f"h = {h:>5}: 前進 {z:.6f}   中心 {c:.6f}")

def kyori(t):
    return t ** 2

# 5秒後の速さ(本当の値は 10)を、広い幅の中心差分で測る
for h in [1, 0.5, 0.1]:
    print(f"h = {h:>3}: 中心 {chuushin(kyori, 5, h):.6f}")
```

## LESSON 16 刻み幅は、小さくするほどいいのか

本書の「LESSON 16 刻み幅は、小さくするほどいいのか」の説明を読みながら、下のコードを ▶ で試してください。

```python
def sanjou(x):
    return x ** 3

# x = 2 での傾き(本当の値は 12)。幅 h をとことん縮めながら測る
for n in range(1, 16):
    h = 0.1 ** n                # 0.1 の n 乗 = 1桁ずつ小さく
    katamuki = (sanjou(2 + h) - sanjou(2 - h)) / (2 * h)
    print(f"h = {h:.0e}: 傾き {katamuki:.12f}")

for n in range(1, 16):
    h = 0.1 ** n
    katamuki = (sanjou(2 + h) - sanjou(2)) / h   # 前進差分に変えた
    print(f"h = {h:.0e}: 傾き {katamuki:.12f}")
```

## LESSON 17 すべての場所の傾きを、一気に描く

本書の「LESSON 17 すべての場所の傾きを、一気に描く」の説明を読みながら、下のコードを ▶ で試してください。

```python
def f(x):
    return x ** 2

x = np.linspace(-3, 3, 61)     # -3 から 3 まで、61 か所の測定点
h = 0.001

# 61 か所ぶんの傾きを、まとめて中心差分で測る
katamuki = (f(x + h) - f(x - h)) / (2 * h)

plt.plot(x, f(x), label="y = x の2乗")
plt.plot(x, katamuki, linestyle="--", label="各点の傾き")
plt.xlabel("x")
plt.ylabel("y")
plt.title("関数と、その傾きのグラフ")
plt.legend()
plt.grid(True)
plt.show()

# 何か所か、数字でも確かめる
for a in [-2, 0, 2]:
    kt = (f(a + h) - f(a - h)) / (2 * h)
    print(f"x = {a:>2} での傾き: {kt:.3f}")

def takasa(t):
    return 20 * t - 5 * t ** 2   # 第2章のボールの高さ

t = np.linspace(0, 4, 61)
h = 0.001
katamuki = (takasa(t + h) - takasa(t - h)) / (2 * h)

plt.plot(t, takasa(t), label="高さ")
plt.plot(t, katamuki, linestyle="--", label="各点の傾き")
plt.axhline(0, color="gray", linewidth=0.8)   # 高さ0の目印
plt.xlabel("時間(秒)")
plt.ylabel("高さ(m)")
plt.title("ボールの高さと、その傾きのグラフ")
plt.legend()
plt.grid(True)
plt.show()
```

## LESSON 18 式が分からなくても、傾きは出せる

本書の「LESSON 18 式が分からなくても、傾きは出せる」の説明を読みながら、下のコードを ▶ で試してください。

```python
x = np.linspace(0, 2 * np.pi, 200)   # 0 から一周(2π)まで、200 か所
h = 0.001
katamuki = (np.sin(x + h) - np.sin(x - h)) / (2 * h)   # 各点の傾き

plt.plot(x, np.sin(x), label="sin x")
plt.plot(x, katamuki, linestyle="--", label="sin x の各点の傾き")
plt.axhline(0, color="gray", linewidth=0.8)
plt.xlabel("x (ラジアン)")
plt.ylabel("y")
plt.title("sin x と、その傾きのグラフ")
plt.legend()
plt.grid(True)
plt.show()

# 出だし(x = 0)の傾きの値も見ておく
hajime = (np.sin(0 + h) - np.sin(0 - h)) / (2 * h)
print(f"x = 0 での傾き: {hajime:.7f}")

plt.plot(x, katamuki, linewidth=5, alpha=0.4, label="sin x の各点の傾き")
plt.plot(x, np.cos(x), linestyle="--", label="cos x")
plt.axhline(0, color="gray", linewidth=0.8)
plt.xlabel("x (ラジアン)")
plt.ylabel("y")
plt.title("傾きのグラフに cos x を重ねる")
plt.legend()
plt.grid(True)
plt.show()

# 式のない、ただの記録: 3時間おきに測った気温
jikoku = [0, 3, 6, 9, 12, 15, 18, 21]     # 時刻
kion = [8, 7, 9, 14, 19, 21, 16, 11]      # 気温(度)

# 各時刻での気温の変化率(1時間あたり何度のペースか)
henka = np.gradient(kion, jikoku)

for j, k, v in zip(jikoku, kion, henka):
    print(f"{j:>2}時: {k:>2}度   1時間あたり {v:+.2f}度")
```
