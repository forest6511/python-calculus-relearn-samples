## LESSON 38 利息の付け方を細かくすると、増え方に上限はあるか

本書の「LESSON 38 利息の付け方を細かくすると、増え方に上限はあるか」の説明を読みながら、下のコードを ▶ で試してください。

```python
for n in [1, 12, 365, 100000]:
    kekka = (1 + 1 / n) ** n       # n回に分けて利息を付けた倍率
    print(f"n={n:>7}: (1 + 1/n)^n = {kekka:.6f}")

print(f"np.e = {np.e:.6f}")

def bibun(f, x, h=0.001):
    return (f(x + h) - f(x - h)) / (2 * h)   # 第5章の傾きの測定器

for x in [0.0, 1.0, 2.0]:
    katamuki = bibun(np.exp, x)    # e^x の傾き
    hontai = np.exp(x)             # e^x の値そのもの
    print(f"x={x}: 傾き {katamuki:.4f}  本体 {hontai:.4f}")
```

## LESSON 39 冷めるコーヒーの温度は、どんな式で下がるか

本書の「LESSON 39 冷めるコーヒーの温度は、どんな式で下がるか」の説明を読みながら、下のコードを ▶ で試してください。

```python
def kabu(t):
    return 2.0 ** t                # t時間後の菌の量(1時間で2倍)

for t in [0.0, 1.0, 3.0]:
    sokudo = bibun(kabu, t)        # 増える速さ(第5章の測定器)
    ima = kabu(t)                  # いまの量
    print(f"t={t}: 増える速さ {sokudo:.4f}  いまの量 {ima:.4f}"
          f"  速さ÷量 {sokudo / ima:.4f}")

def coffee(t):
    return 20 + 70 * np.exp(-0.1 * t)   # 室温20度、初期90度の冷却

t = np.linspace(0, 30, 7)
for tt, oo in zip(t, coffee(t)):
    print(f"t={tt:>4.0f}分: {oo:.2f}度")

t = np.linspace(0, 60, 200)
plt.plot(t, coffee(t))
plt.axhline(20, color="gray", linestyle=":", label="室温 20度")
plt.xlabel("時間(分)")
plt.ylabel("温度(度)")
plt.title("コーヒーが冷めていくようす")
plt.legend()
plt.grid(True)
plt.show()

for t in [0.0, 10.0, 20.0]:
    sokudo = bibun(coffee, t)      # 冷める速さ(負の値)
    sa = coffee(t) - 20            # 室温との差
    print(f"t={t}: 冷める速さ {sokudo:+.4f}  室温との差 {sa:.4f}"
          f"  速さ÷差 {sokudo / sa:+.4f}")
```

## LESSON 40 急成長するデータを、どうすれば読みやすく描けるか

本書の「LESSON 40 急成長するデータを、どうすれば読みやすく描けるか」の説明を読みながら、下のコードを ▶ で試してください。

```python
t = np.linspace(0, 10, 11)
y = 2.0 ** t

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax1.plot(t, y, marker="o")
ax1.set_title("ふつうの目盛り")
ax1.set_xlabel("日")
ax1.grid(True)

ax2.plot(t, y, marker="o")
ax2.set_yscale("log")               # 縦軸を対数の目盛りに変える
ax2.set_title("対数の目盛り")
ax2.set_xlabel("日")
ax2.grid(True)
plt.show()

x = np.linspace(1, np.e, 100001)
menseki = np.trapezoid(1 / x, x)    # 1からeまでの 1/x の面積

print(f"1 から e まで 1/x の面積: {menseki:.6f}")
print(f"np.log(np.e)           : {np.log(np.e):.6f}")

x = np.linspace(1, 2, 100001)
menseki = np.trapezoid(1 / x, x)

print(f"1 から 2 まで 1/x の面積: {menseki:.6f}")
print(f"np.log(2)              : {np.log(2):.6f}")

import sympy

x_moji = sympy.Symbol("x")
print(sympy.integrate(1 / x_moji, x_moji))    # 1/x の不定積分
print(sympy.diff(sympy.log(x_moji), x_moji))  # log x の微分
```
