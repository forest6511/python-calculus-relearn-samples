## LESSON 34 面積を積み上げたグラフの傾きは、何になるか

本書の「LESSON 34 面積を積み上げたグラフの傾きは、何になるか」の説明を読みながら、下のコードを ▶ で試してください。

```python
def bouru(t):
    return 20 - 10 * t              # 第2章のボールの速度(m/秒)

t = np.linspace(0, 4, 401)
tsumiage = []                       # 面積の積み上げ(第7章の再演)
for k in range(len(t)):
    made = t[: k + 1]               # 先頭からいまの時刻まで
    tsumiage.append(np.trapezoid(bouru(made), made))
tsumiage = np.array(tsumiage)

katamuki = np.gradient(tsumiage, t) # 積み上げのグラフの傾き

print(f"t=1 での積み上げの傾き: {katamuki[100]:+.2f}   速度: {bouru(1):+.2f}")
print(f"t=3 での積み上げの傾き: {katamuki[300]:+.2f}   速度: {bouru(3):+.2f}")

plt.plot(t, katamuki, label="積み上げのグラフの傾き")
plt.plot(t, bouru(t), linestyle="--", label="速度 20 − 10t")
plt.xlabel("時間(秒)")
plt.ylabel("速度(m/秒)")
plt.title("積み上げの傾きは、元の速度に戻る")
plt.legend()
plt.grid(True)
plt.show()

def nijou(x):
    return x ** 2

x = np.linspace(0, 1, 401)
S = []
for k in range(len(x)):
    S.append(np.trapezoid(nijou(x[: k + 1]), x[: k + 1]))
katamuki2 = np.gradient(np.array(S), x)

print(f"x=0.5 での積み上げの傾き: {katamuki2[200]:.4f}   もとの関数: {nijou(0.5):.4f}")
print(f"x=0.8 での積み上げの傾き: {katamuki2[320]:.4f}   もとの関数: {nijou(0.8):.4f}")
```

## LESSON 35 速度の記録だけで、投げた高さまで分かるか

本書の「LESSON 35 速度の記録だけで、投げた高さまで分かるか」の説明を読みながら、下のコードを ▶ で試してください。

```python
def takasa_jimen(t):
    return 20 * t - 5 * t ** 2      # 地面から投げた高さ

def takasa_dai(t):
    return 5 + 20 * t - 5 * t ** 2  # 高さ5mの台から投げた高さ

def doukansuu(f, x, h=0.001):
    return (f(x + h) - f(x - h)) / (2 * h)   # 第5章の測定器

for t in [1.0, 3.0]:
    jimen = doukansuu(takasa_jimen, t)
    dai = doukansuu(takasa_dai, t)
    print(f"t={t}: 地面の傾き {jimen:+.2f}   台の傾き {dai:+.2f}")

t = np.linspace(0, 4, 401)
sokudo = doukansuu(takasa_dai, t)   # 台の高さの傾き(速度)の列
modoshi = []
for k in range(len(t)):
    modoshi.append(np.trapezoid(sokudo[: k + 1], t[: k + 1]))

plt.plot(t, modoshi, label="傾きを積み上げて戻した高さ")
plt.plot(t, takasa_dai(t), linestyle="--", label="本物の高さ 5 + 20t − 5t²")
plt.xlabel("時間(秒)")
plt.ylabel("高さ(m)")
plt.title("戻した高さは、本物より5m低い")
plt.legend()
plt.grid(True)
plt.show()

print(f"t=2 で戻した高さ: {modoshi[200]:.2f} m   本物: {takasa_dai(2):.2f} m")

naoshi = np.array(modoshi) + takasa_dai(0)   # 出発点の高さを足す
print(f"t=2: 直した高さ {naoshi[200]:.2f} m   本物 {takasa_dai(2):.2f} m")
```

## LESSON 36 短冊を1本も足さずに、面積は出せるか

本書の「LESSON 36 短冊を1本も足さずに、面積は出せるか」の説明を読みながら、下のコードを ▶ で試してください。

```python
def F(x):
    return x ** 3 / 3               # 微分すると x² に戻るはずの関数

print(f"F の x=0.5 での傾き: {doukansuu(F, 0.5):.4f}")
print(f"x=0.5 での x²      : {nijou(0.5):.4f}")

x = np.linspace(0, 1, 101)
daikei = np.trapezoid(nijou(x), x)

print(f"F(1) − F(0) = {F(1) - F(0):.10f}")
print(f"1 ÷ 3       = {1 / 3:.10f}")
print(f"台形100個   = {daikei:.10f}")

def F_sokudo(t):
    return 3 * t ** 2 / 2           # 微分すると 3t に戻る関数

print(f"進んだ距離: {F_sokudo(10) - F_sokudo(0):.1f} m")
```

## LESSON 37 微分すると x² になる関数は、どうやって探すか

本書の「LESSON 37 微分すると x² になる関数は、どうやって探すか」の説明を読みながら、下のコードを ▶ で試してください。

```python
def gyaku_beki(x, n):
    return x ** (n + 1) / (n + 1)   # 肩を1増やして、増えた肩で割る

for n in [1, 2, 3]:
    x = np.linspace(0, 2, 201)
    suuchi = np.trapezoid(x ** n, x)
    shiki = gyaku_beki(2, n) - gyaku_beki(0, n)
    print(f"x^{n} の 0〜2 の面積: 台形則 {suuchi:.4f}   公式 {shiki:.4f}")

import sympy

x_moji = sympy.Symbol("x")          # 文字のままの x を作る
print(sympy.integrate(x_moji ** 2, x_moji))
print(sympy.integrate(x_moji ** 2, (x_moji, 0, 1)))

t_moji = sympy.Symbol("t")
print(sympy.integrate(20 - 10 * t_moji, t_moji))
```
