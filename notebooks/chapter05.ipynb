## LESSON 19 x³ の傾きは、測らなくても分かるか

本書の「LESSON 19 x³ の傾きは、測らなくても分かるか」の説明を読みながら、下のコードを ▶ で試してください。

```python
# 傾きの測定器(中心差分)。この章ではこれを使い回す
def doukansuu(f, x, h=0.001):
    return (f(x + h) - f(x - h)) / (2 * h)

def sanjou(x):
    return x ** 3

for a in [-2, -1, 0, 1, 2]:
    print(f"x = {a:>2} での傾き: {doukansuu(sanjou, a):.6f}")

x = np.linspace(-2, 2, 81)

plt.plot(x, doukansuu(sanjou, x), linewidth=5, alpha=0.4,
         label="x の3乗の各点の傾き")
plt.plot(x, 3 * x ** 2, linestyle="--", label="3 × x の2乗")
plt.xlabel("x")
plt.ylabel("傾き")
plt.title("x の3乗の傾きのグラフの正体")
plt.legend()
plt.grid(True)
plt.show()

def yonjou(x):
    return x ** 4

for a in [-2, -1, 0, 1, 2]:
    print(f"x = {a:>2} での傾き: {doukansuu(yonjou, a):.6f}")

def gojou(x):
    return x ** 5

print(f"x =  1 での傾き: {doukansuu(gojou, 1):.6f}")
```

## LESSON 20 足し算と定数倍は、バラして微分してよい

本書の「LESSON 20 足し算と定数倍は、バラして微分してよい」の説明を読みながら、下のコードを ▶ で試してください。

```python
def nijou(x):
    return x ** 2

def nijou_3bai(x):
    return 3 * x ** 2

for a in [-1, 1, 2]:
    moto = doukansuu(nijou, a)
    bai = doukansuu(nijou_3bai, a)
    print(f"x = {a:>2}: もとの傾き {moto:8.4f}   3倍した関数の傾き {bai:8.4f}")

def sanjou(x):
    return x ** 3

def wa(x):
    return x ** 2 + x ** 3

for a in [-1, 1, 2]:
    barabara = doukansuu(nijou, a) + doukansuu(sanjou, a)
    matomete = doukansuu(wa, a)
    print(f"x = {a:>2}: バラして足す {barabara:8.4f}   まとめて測る {matomete:8.4f}")

def takasa(t):
    return 20 * t - 5 * t ** 2

for t in [0, 1, 2, 3]:
    kisoku = 20 - 10 * t              # 規則の組み立てで出した傾き
    jissoku = doukansuu(takasa, t)    # 数値で測った傾き
    print(f"t = {t}: 規則で {kisoku:6.2f}   実測 {jissoku:6.2f}")
```

## LESSON 21 掛け算の微分は、傾きの掛け算ではない

本書の「LESSON 21 掛け算の微分は、傾きの掛け算ではない」の説明を読みながら、下のコードを ▶ で試してください。

```python
def f(x):
    return x ** 2

def g(x):
    return x ** 3

def seki(x):
    return x ** 2 * x ** 3     # 掛け算した関数(x の5乗)

a = 1
yosou = doukansuu(f, a) * doukansuu(g, a)    # 予想: 傾き × 傾き
jissoku = doukansuu(seki, a)                 # ぜんたいを実測
print(f"傾き × 傾き の予想: {yosou:.4f}")
print(f"実測: {jissoku:.4f}")

for a in [1, 2]:
    obi = doukansuu(f, a) * g(a) + f(a) * doukansuu(g, a)
    jissoku = doukansuu(seki, a)
    print(f"x = {a}: 帯2本の合計 {obi:8.4f}   実測 {jissoku:8.4f}")

def nami(x):
    return x ** 2 * np.sin(x)    # まとめられない掛け算

a = 1
obi = 2 * a * np.sin(a) + a ** 2 * np.cos(a)   # 帯2本の式で
jissoku = doukansuu(nami, a)                   # 実測で
print(f"帯2本の式で: {obi:.4f}")
print(f"実測で: {jissoku:.4f}")
```

## LESSON 22 入れ子の関数は、速さの掛け算で微分する

本書の「LESSON 22 入れ子の関数は、速さの掛け算で微分する」の説明を読みながら、下のコードを ▶ で試してください。

```python
def soto(u):
    return u ** 2          # 外側: 2乗する

def naka(x):
    return 2 * x + 1       # 内側: 2倍して1を足す

def ireko(x):
    return soto(naka(x))   # 入れ子: (2x + 1) の2乗

a = 1
print(f"入れ子ぜんたいの傾き: {doukansuu(ireko, a):.4f}")
print(f"外の傾き(u = {naka(a)} の場所で): {doukansuu(soto, naka(a)):.4f}")
print(f"中の傾き: {doukansuu(naka, a):.4f}")

def soto3(u):
    return u ** 3

def naka2(x):
    return x ** 2 + 1

def ireko2(x):
    return soto3(naka2(x))

a = 1
print(f"入れ子ぜんたいの傾き: {doukansuu(ireko2, a):.4f}")
print(f"外の傾き × 中の傾き: {doukansuu(soto3, naka2(a)) * doukansuu(naka2, a):.4f}")
```

## LESSON 23 傾きが自分自身になる、特別な関数

本書の「LESSON 23 傾きが自分自身になる、特別な関数」の説明を読みながら、下のコードを ▶ で試してください。

```python
def bai_2(x):
    return 2.0 ** x        # 2 の x乗

def bai_3(x):
    return 3.0 ** x        # 3 の x乗

x = np.linspace(0, 3, 61)
hi_2 = doukansuu(bai_2, x) / bai_2(x)     # 傾き ÷ 本体(61か所ぶん)
hi_3 = doukansuu(bai_3, x) / bai_3(x)
print(f"2 の x乗: 傾き ÷ 本体 は {hi_2.min():.6f} 〜 {hi_2.max():.6f}")
print(f"3 の x乗: 傾き ÷ 本体 は {hi_3.min():.6f} 〜 {hi_3.max():.6f}")

h = 0.001
for soko in [2.0, 2.5, 2.7, 2.72, 3.0]:
    katamuki = (soko ** (1 + h) - soko ** (1 - h)) / (2 * h)   # x=1 での傾き
    hiritsu = katamuki / soko           # 本体(soko の1乗 = soko)で割る
    print(f"底 {soko:>4}: 傾き ÷ 本体 = {hiritsu:.6f}")

print(f"特別な底 e: {np.e:.6f}")
print(f"e の x乗 の x = 1 での傾き: {doukansuu(np.exp, 1.0):.6f}")
print(f"e の x乗 の x = 1 での本体: {np.exp(1.0):.6f}")

for a in [0.5, 1.0, 2.0, 4.0]:
    print(f"x = {a}: log x の傾き {doukansuu(np.log, a):.4f}   1 ÷ x {1 / a:.4f}")

print(f"cos の x = 1 での傾き: {doukansuu(np.cos, 1.0):.6f}")
print(f"マイナス sin(1) の値: {-np.sin(1.0):.6f}")
```
