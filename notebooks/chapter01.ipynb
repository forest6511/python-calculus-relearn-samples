## LESSON 01 「xが1増えると、いくつ増える?」を絵にする

```python
# 貯金の合計 = 1000円 × 月数 + 最初の3000円
month = np.linspace(0, 12, 13)   # 0か月から12か月まで
saving = 1000 * month + 3000     # 毎月1000円ずつ増える

plt.plot(month, saving, marker="o")   # 点をつないで線にする
plt.xlabel("月数")
plt.ylabel("貯金の合計(円)")
plt.title("毎月1000円ずつ貯金する")
plt.grid(True)
plt.show()

# 具体的な値を確認する
for m in [0, 1, 2, 3]:
    print(f"{m}か月後: {int(1000 * m + 3000)}円")

saving = 1500 * month + 3000   # ペースを1500円に上げた
```

## LESSON 02 傾きを、2つの点から数える

本書の「LESSON 02 傾きを、2つの点から数える」の説明を読みながら、下のコードを ▶ で試してください。

```python
# 2つの点(月数, 貯金額)
x1, y1 = 2, 5000
x2, y2 = 5, 8000

# 傾き = 縦の上がり ÷ 横の進み
delta_y = y2 - y1        # 縦の上がり
delta_x = x2 - x1        # 横の進み
katamuki = delta_y / delta_x

print(f"縦の上がり: {delta_y}円")
print(f"横の進み: {delta_x}か月")
print(f"傾き: {katamuki}円/月")

x2, y2 = 5, 9500   # 2つ目の点を変えた
```

## LESSON 03 まっすぐでない線、放物線を描く

本書の「LESSON 03 まっすぐでない線、放物線を描く」の説明を読みながら、下のコードを ▶ で試してください。

```python
# -3 から 3 までの x を細かく用意する
x = np.linspace(-3, 3, 100)
y = x ** 2          # それぞれの x を 2 乗する

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y = x の2乗")
plt.title("放物線 y = x の2乗")
plt.grid(True)
plt.axhline(0, color="gray", linewidth=0.8)  # 横軸の線
plt.axvline(0, color="gray", linewidth=0.8)  # 縦軸の線
plt.show()

# いくつかの値を確認する
for xi in [-2, -1, 0, 1, 2]:
    print(f"x = {xi:>2} のとき y = {xi ** 2}")

y = x ** 2 + 2   # 全体を上に2持ち上げる
```

## LESSON 04 放物線の傾きは、場所で変わる

本書の「LESSON 04 放物線の傾きは、場所で変わる」の説明を読みながら、下のコードを ▶ で試してください。

```python
def f(x):
    return x ** 2        # 放物線 y = x の2乗

# 区間ごとの傾き = (yの増加) ÷ (xの増加)
def katamuki(x_start, x_end):
    return (f(x_end) - f(x_start)) / (x_end - x_start)

print(f"x=0 から 1 の傾き: {katamuki(0, 1)}")
print(f"x=1 から 2 の傾き: {katamuki(1, 2)}")
print(f"x=2 から 3 の傾き: {katamuki(2, 3)}")

print(f"x=3 から 4 の傾き: {katamuki(3, 4)}")
```

## LESSON 05 関数を「入れると出てくる箱」として捉える

本書の「LESSON 05 関数を「入れると出てくる箱」として捉える」の説明を読みながら、下のコードを ▶ で試してください。

```python
# 「2乗する箱」に f という名前をつける
def f(x):
    return x ** 2      # 入ってきた x を 2 乗して返す

# 箱にいろいろ入れてみる
print(f"3 を入れると: {f(3)}")
print(f"5 を入れると: {f(5)}")
print(f"-4 を入れると: {f(-4)}")

# 別の箱「2倍して1足す」も作ってみる
def g(x):
    return 2 * x + 1

print(f"g に 10 を入れると: {g(10)}")

def h(x):
    return 3 * x - 2

print(f"h に 8 を入れると: {h(8)}")
```
