## LESSON 10 半分ずつ進むと、どこへ行き着くのか

本書の「LESSON 10 半分ずつ進むと、どこへ行き着くのか」の説明を読みながら、下のコードを ▶ で試してください。

```python
# 壁までの残り 2m を、毎回「残りの半分」だけ進む
ichi = 0                       # いまの位置(m)。壁は 2m 先にある
for kaisuu in range(1, 11):    # 1歩目から10歩目まで繰り返す
    ichi = ichi + (2 - ichi) / 2   # 残り (2 - ichi) の半分だけ前へ
    print(f"{kaisuu:>2}歩目: 位置 {ichi:.6f} m   壁まで {2 - ichi:.6f} m")

ichi = 0
for kaisuu in range(1, 6):     # 5歩目まで見れば十分
    ichi = ichi + (2 - ichi) * 0.9   # 残りの9割だけ前へ
    print(f"{kaisuu:>2}歩目: 位置 {ichi:.6f} m   壁まで {2 - ichi:.6f} m")
```

## LESSON 11 0÷0 に見える式が、1つの値に落ち着く理由

本書の「LESSON 11 0÷0 に見える式が、1つの値に落ち着く理由」の説明を読みながら、下のコードを ▶ で試してください。

```python
def f(t):
    return (t ** 2 - 25) / (t - 5)   # t = 5 だけ 0÷0 になる式

# 5 の両側から近づく
for t in [4.9, 4.99, 4.999, 5.001, 5.01, 5.1]:
    print(f"t = {t:>5} のとき: {f(t):.4f}")

def f(t):
    return (t ** 2 - 16) / (t - 4)   # 25 と 5 を、16 と 4 に変えた

for t in [3.9, 3.99, 3.999, 4.001, 4.01, 4.1]:
    print(f"t = {t:>5} のとき: {f(t):.4f}")
```

## LESSON 12 sin(x) ÷ x は、どこへ向かうのか

本書の「LESSON 12 sin(x) ÷ x は、どこへ向かうのか」の説明を読みながら、下のコードを ▶ で試してください。

```python
# x を 0 に近づけながら、sin(x) ÷ x を計算する(x はラジアン)
for x in [0.5, 0.1, 0.01, 0.001]:
    print(f"x = {x:>5}: sin(x) = {np.sin(x):.7f} → 割ると {np.sin(x) / x:.7f}")

# sin(x) ÷ x のグラフも描いて、0 の近くのようすを見る
x = np.linspace(-10, 10, 500)   # 500個の点は 0 ちょうどを踏まない
y = np.sin(x) / x

plt.plot(x, y)
plt.axhline(1, color="gray", linestyle="--", linewidth=0.8)  # 高さ1の目印
plt.xlabel("x (ラジアン)")
plt.ylabel("sin(x) ÷ x")
plt.title("sin(x) ÷ x のグラフ")
plt.grid(True)
plt.show()

for x in [0.1, 0.01, 0.001]:
    print(f"x = {x:>5}: sin(2x) ÷ x = {np.sin(2 * x) / x:.7f}")
```

## LESSON 13 行き先が1つに決まらないこともある

本書の「LESSON 13 行き先が1つに決まらないこともある」の説明を読みながら、下のコードを ▶ で試してください。

```python
# 重さ w kg の荷物の送料(円)
def souryou(w):
    if w <= 1:
        return 800     # 1kg までは 800円
    else:
        return 1200    # 1kg を超えると 1200円

# 1kg の両側から近づく
for w in [0.9, 0.99, 0.999, 1.001, 1.01, 1.1]:
    print(f"w = {w:>5} kg のとき: {souryou(w)}円")

def souryou(w):
    if w <= 1:
        return 800
    else:
        return 800 + 400 * (w - 1)   # 超えたぶんだけ加算する方式に変えた

for w in [0.9, 0.99, 0.999, 1.001, 1.01, 1.1]:
    print(f"w = {w:>5} kg のとき: {souryou(w):.1f}円")
```
